```python
 from bing_image_downloader import downloader
```

```python
downloader.download("uzbek pilaf plov food",
                    limit=100,
                    output_dir="/content/pilaf_images",
                    adult_filter_off=True,
                    force_replace=False,
                    timeout=60)
```

```python
from google.colab import drive
drive.mount('/content/driv')
```

```python
import os, shutil
source_dir = "/content/pilaf_images/uzbek pilaf plov food"
target_dir = "/content/driv/MyDrive/food101_dataset/pilaf"

os.makedirs(target_dir, exist_ok=True)

for filename in os.listdir(source_dir):
  src_file = os.path.join(source_dir, filename)
  dst_file = os.path.join(target_dir, filename)
  shutil.move(src_file, dst_file)
```
